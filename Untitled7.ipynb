```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn .cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
df=pd.read_csv(r"C:\Users\arulr\OneDrive\Desktop\REPETITION\customer_purchase_history.csv")
print(df.head())
```

```python
X=df[['Quantity','FinalPrice','Profit']]
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)

inertia=[]
for k in range(1,11):
    km=KMeans(n_clusters=k, random_state=42)
    km.fit(X_scaled)
    inertia.append(km.inertia_)

plt.figure(figsize=(8,5))
plt.plot(range(1,11),inertia,marker='o',color='blue')
plt.xlabel('Number of Clusters(K)')
plt.ylabel('inertia')
plt.title('Elbow Method - Finding Best K')
plt.show()
```

```python
km=KMeans(n_clusters=3,random_state=42)
df['cluster']=km.fit_predict(X_scaled)

print(df['cluster'].value_counts())

print(df.groupby('cluster')[['Quantity','FinalPrice','Profit']].mean())
```

```python
plt.figure(figsize=(8,5))
colors=['blue','green','red']
labels=['Budget','Regular','Premium']

for i in range(3):
    cluster_data=df[df['cluster']==i]
    plt.scatter(cluster_data['Quantity'],
    cluster_data['FinalPrice'],
    c=colors[i],label=labels[i],alpha=0.5)

plt.xlabel('Quantity')
plt.ylabel('Final Price')
plt.title('Customer Segments')
plt.legend()
plt.show()
```
